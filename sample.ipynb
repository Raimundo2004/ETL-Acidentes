```python
#FILTRANDO ACIDENTES QUE ACONTECEU EM SP

db_SP_2026 = db_2026[db_2026['uf_acidente']=='SP']
db_SP_2022 = db_2022[db_2022['uf']=='SP']
db_SP_2021 = db_2021[db_2021['uf']=='SP']
db_SP_2020 = db_2020[db_2020['uf']=='SP']
db_SP_2019 = db_2019[db_2019['uf']=='SP']
db_SP_2024 = db_2024[db_2024['uf']=='SP']
db_SP_2023 = db_2023[db_2023['uf']=='SP']
db_SP_2025 = db_2025[db_2025['uf']=='SP']
db_SP_2018 = db_2018[db_2018['uf']=='SP']
db_SP_2017 = db_2017[db_2017['uf']=='SP']
```

```python
#COLOCANDO OS ANOS QUE ACONTECEU OS ACIDENTES

db_SP_2017['ano'] = 2017
db_SP_2018['ano'] = 2018
db_SP_2019['ano'] = 2019
db_SP_2020['ano'] = 2020
db_SP_2021['ano'] = 2021
db_SP_2022['ano'] = 2022
db_SP_2023['ano'] = 2023
db_SP_2024['ano'] = 2024
db_SP_2025['ano'] = 2025
db_SP_2026['ano'] = 2026
```

```python
# RENOMEANDO AS COLUNAS DA TABELA DE 2026, PARA PODER CONCATENAR COM AS OUTRAS

db_SP_2026 = db_SP_2026.rename(columns={'cd_bat': 'id',
    'id_envolvido': 'pesid',
    'uf_acidente': 'uf',
    'rodovia': 'br',
    'cond_meteorologica': 'condicao_metereologica',
    'estrutura_viaria': 'tracado_via',
    'local_urbanizado': 'uso_solo',
    'ano_fabricacao': 'ano_fabricacao_veiculo',
    'qtde_ileso': 'ilesos',
    'qtde_lesoes_leves': 'feridos_leves',
    'qtde_lesoes_graves': 'feridos_graves',
    'qtde_mortos': 'mortos',
    'Sigla da Superintendência': 'regional',
    'Sigla da Delegacia': 'delegacia',
    'Sigla da Unidade Operacional': 'uop'})
db_SP_2026
```

```python
##JUNTANDO AS TABELAS

db_SP = pd.concat([db_SP_2017,db_SP_2018,db_SP_2019,db_SP_2020,db_SP_2021,db_SP_2022,db_SP_2023,db_SP_2025,db_SP_2026])

db_SP
```

```python
## PREENCHENDO VALORES NULOS

db_SP['sexo'] = db_SP['sexo'].fillna('Não Informado')
db_SP['sexo'] = db_SP['sexo'].replace('Ignorado','Não Informado')
db_SP['ilesos'] = db_SP['ilesos'].fillna(-1)
db_SP['feridos_leves'] = db_SP['feridos_leves'].fillna(-1)
db_SP['feridos_graves'] = db_SP['feridos_graves'].fillna(-1)
db_SP['mortos'] = db_SP['mortos'].fillna(-1)
db_SP['estado_fisico'] = db_SP['estado_fisico'].fillna('Não Informado')
db_SP['tipo_envolvido'] = db_SP['tipo_envolvido'].fillna('Não Informado')
db_SP['regional'] = db_SP['regional'].fillna('Não Informado')
db_SP['delegacia'] = db_SP['delegacia'].fillna('Não Informado')
db_SP['uop'] = db_SP['uop'].fillna('Não Informado')
```

```python
# PREENCHENDO VALORES NULOS DA IDADE

media_idade = db_SP.groupby(['tipo_acidente','municipio','tipo_envolvido','sexo'])['idade'].mean()
db_SP['idade'] = db_SP['idade'].fillna(db_SP.groupby(['tipo_acidente','municipio','tipo_envolvido','sexo'])['idade'].transform('mean'))
db_SP['idade'] = db_SP['idade'].fillna(db_SP.groupby(['sexo'])['idade'].transform('mean'))

# RETIRANDO VALORES NULOS DA IDADE

Q1 = db_SP['idade'].quantile(0.25)
Q3 = db_SP['idade'].quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR
mediana_valida = db_SP[(db_SP['idade'] >= limite_inferior) & (db_SP['idade'] <= limite_superior)]['idade'].median()
condicao_outlier = (db_SP['idade'] < limite_inferior) | (db_SP['idade'] > limite_superior)
db_SP['idade'] = np.where(condicao_outlier, mediana_valida, db_SP['idade'])
```

```python
# REMOVENDO OS ACENTOS, PARA FICAR LIMPO NO BANCO DE DADOS
def remover_acentos(texto):
    if isinstance(texto, str):
        # Normaliza decompondo a letra e o acento
        nfkd = unicodedata.normalize('NFKD', texto)
        # Mantém apenas a letra e descarta o acento
        return "".join([c for c in nfkd if not unicodedata.combining(c)])
    return texto

db_SP.columns = [
    remover_acentos(col).lower().replace(' ', '_').replace('-', '_')
    for col in db_SP.columns
]

colunas_texto = db_SP.select_dtypes(include=['object', 'string']).columns

for col in colunas_texto:
    db_SP[col] = db_SP[col].apply(remover_acentos)
```

```python
# CONVERTENDO OS TIPOS DOS VALORES

db_SP['id'] = db_SP['id'].astype('Int64')
db_SP['pesid'] = db_SP['pesid'].astype('Int64')
db_SP['data_inversa'] = pd.to_datetime(db_SP['data_inversa'])
db_SP['br'] = db_SP['br'].astype('Int64')
db_SP['km'] = db_SP['km'].replace(',','.')
db_SP['km'] = pd.to_numeric(db_SP['km'], errors='coerce')
db_SP['ordem_tipo_acidente'] = db_SP['ordem_tipo_acidente'].astype('Int64')
db_SP['id_veiculo'] = db_SP['id_veiculo'].astype('Int64')
db_SP['ano_fabricacao_veiculo'] = db_SP['ano_fabricacao_veiculo'].astype('Int64')
db_SP['idade'] = pd.to_numeric(db_SP['idade'], errors='coerce')
db_SP['idade'] = np.trunc(db_SP['idade']).astype('Int64')
db_SP['ilesos'] = db_SP['ilesos'].astype('Int64')
db_SP['feridos_leves']= db_SP['feridos_leves'].astype('Int64')
db_SP['feridos_graves'] = db_SP['feridos_graves'].astype('Int64')
db_SP['mortos'] = db_SP['mortos'].astype('Int64')
db_SP['ano'] = db_SP['ano'].astype('Int64')
```
